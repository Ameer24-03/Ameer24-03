```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
import joblib
def load_data(csv_file):
    data = pd.read_csv('/content/final_dataset.csv')
    return data
def preprocess_data(data):
    X = data[['beds', 'baths', 'size', 'zip_code']]
    y = data['price']
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled, y, scaler
def train_model(X, y):
    model = LinearRegression()
    model.fit(X, y)
    return model
def predict_price(model, scaler, beds, baths, size, zip_code):
    input_data = [[beds, baths, size, zip_code]]
    input_data_scaled = scaler.transform(input_data)
    prediction = model.predict(input_data_scaled)
    return prediction[0]
def save_model(model, scaler, model_file, scaler_file):
    joblib.dump(model, model_file)
    joblib.dump(scaler, scaler_file)
def load_model(model_file, scaler_file):
    model = joblib.load(model_file)
    scaler = joblib.load(scaler_file)
    return model, scaler
def main():
    data = load_data('/content/final_dataset.csv')
    X, y, scaler = preprocess_data(data)
    model = train_model(X, y)
    save_model(model, scaler, 'house_price_model.pkl', 'scaler.pkl')
    beds = int(input("Enter number of bedrooms: "))
    baths = int(input("Enter number of bathrooms: "))
    size = float(input("Enter size of the house in sqft: "))
    zip_code = int(input("Enter zip code: "))
    predicted_price = predict_price(model, scaler, beds, baths, size, zip_code)
    print(f"Predicted House Price: ${predicted_price:,.2f}")
if __name__ == "__main__":
    main()
```
